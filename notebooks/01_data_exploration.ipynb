```python
from scipy.stats import chi2_contingency

variables = [
    "Type of Admission",
    "Illness_Severity",
    "Age"
]

for variable in variables:
    table = pd.crosstab(df[variable], df["Stay_Days"])
    
    chi2, p_value, dof, expected = chi2_contingency(table)
    
    print(f"\n{variable}")
    print(f"Chi-square: {chi2:.4f}")
    print(f"p-value: {p_value:.6e}")
    print(f"Degrees of freedom: {dof}")
```

```python
def cramers_v(table):
    chi2, _, _, _ = chi2_contingency(table)
    
    n = table.to_numpy().sum()
    r, k = table.shape
    
    return np.sqrt(
        chi2 / (n * min(r - 1, k - 1))
    )


for variable in variables:
    table = pd.crosstab(
        df[variable],
        df["Stay_Days"]
    )
    
    v = cramers_v(table)
    
    print(f"{variable}: Cramér's V = {v:.4f}")
```

```python
from scipy.stats import chi2_contingency
import numpy as np
import pandas as pd


def cramers_v(table):
    chi2, _, _, _ = chi2_contingency(table)

    n = table.to_numpy().sum()
    r, k = table.shape

    return np.sqrt(
        chi2 / (n * min(r - 1, k - 1))
    )


results = []

for variable in categorical_variables:
    table = pd.crosstab(
        df[variable],
        df["Stay_Days"]
    )

    chi2, p_value, dof, expected = chi2_contingency(table)
    v = cramers_v(table)

    results.append({
        "Variable": variable,
        "Chi2": chi2,
        "p_value": p_value,
        "Cramers_V": v,
        "Degrees_of_Freedom": dof
    })


association_results = (
    pd.DataFrame(results)
    .sort_values("Cramers_V", ascending=False)
    .reset_index(drop=True)
)

association_results
```

```python
n = len(df)
k = df["Stay_Days"].nunique()

kruskal_results = []

for variable in numeric_variables:
    groups = [
        group[variable].dropna().values
        for _, group in df.groupby("Stay_Days", observed=True)
    ]

    H, p_value = kruskal(*groups)

    epsilon_squared = (H - k + 1) / (n - k)

    kruskal_results.append({
        "Variable": variable,
        "H": H,
        "p_value": p_value,
        "Epsilon_Squared": epsilon_squared
    })

kruskal_results = pd.DataFrame(kruskal_results)

display(kruskal_results)
```
